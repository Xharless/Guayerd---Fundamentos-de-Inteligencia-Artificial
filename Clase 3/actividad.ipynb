# Cuantos años tienes 


```python
nombre = input("Ingresa tu nombre: ")
año_de_nacimieno = input("Ingresa tu año de nacimiento: ")
print(type(año_de_nacimieno))
año_de_nacimieno = int(año_de_nacimieno)
edad = 2025 - año_de_nacimieno
print(type(edad))
if(edad >= 18):
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")

print(f"Hola {nombre}, tienes {edad} años")
```

# Lista de compras

```python
diccionario = {}
i = 0
while(i<3):
    producto = input("Ingresa el nombre del producto: ")
    precio = input("Ingresa el precio del producto: ")
    precio = float(precio)
    diccionario[producto] = precio
    i += 1

print(diccionario)

total = sum(diccionario.values())
print(f"El total de la compra es: {total}")

producto_mas_caro = max(diccionario, key=diccionario.get) # entrega la llave del valor maximo
print(f"El producto más caro es: {producto_mas_caro} con un precio de {diccionario[producto_mas_caro]}")
```
